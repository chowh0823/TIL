```python
# 함수는 반복적인 작업을 수행할 때 함수를 만든다.
# 재사용 가능한 코드의 묶음을 구성할 수 있기 때문이다.
```

```python
# 함수의 기본적인 구조

def greetings(user_name, gender, job='student'): # 매개변수(parameter), 기본 인자 값
    """
    user_name 이라는 parameter를 받아 콘솔에 출력하는 함수
    """
    print(f"안녕하세요. 제 이름은 {user_name}입니다.")
    
# 함수 호출 및 인자 전달
greetings("조원혁", "male") # 인자(argument) 위치 인자 전달 방식
greetings(user_name="조원혁", gender="male") # 인자(argument) 키워드 인자 전달 방식
# 단, 키워드 인자는 반드시 위치 인자 뒤에 위치해야된다.
```

```python
# 유연, 다양한 인자 전달 방식

# 정해지지 않은 인자를 넘기고 싶을 때 / 파라미터를 받고 싶을 때
# 함수를 정의할 때 매개변수 앞에 '*' 붙여 사용하면 여러개의 인자를 'tuple' 묶어 처리해준다

def calcuate_average(*numbers):
    # numbers가 0개 이상일 때만 계산을 수행한다.
    return sum(numbers) / len(numbers)

print(calcuate_average(100, 40, 56, 45,88))

# '**' 붙여 여러개의 인자를 'dictionary'로 묶어 처리
def user_detail(**details):
    print(details)
    
user_detail(user_name="조원혁", gender="male", city = "seoul")

# 함수를 선언할 때 매개변수 선은의 순서!: 위치, 기본(default), 가변, 가변 키워드
```

```python
# 파이썬에서의 함수 호출
# 재귀함수는 어렵다.

def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)
    
factorial(3)
```

1. `factorial(3)` 호출
   - n(3)은 1이 아니므로, else 블럭으로 감
   - `n * factorial(n-1)`를 계산해야 함을 알려요. `factorial(2)`의 결과가 나올 때까지 `factorial(3)`은 잠시 대기합니다.
2. `factorial(2)` 호출
   - `factorial(3)` 내부에서 `factorial(2)`가 호출 됐어요. n은 2입니다.
     - n 은 1이 아니므로 else 블럭으로 감
     - `n * factorial(n-1)`를 계산해야 함을 알려요. `factorial(1)`의 결과가 나올 때까지 `factorial(2)`은 잠시 대기합니다.
3. `factorial(1)` 호출 (base case 도달) -> **재귀 함수는 반드시 base case 종료 조건이 필요하다.**
   - `factorial(2)` 내부에서 `factorial(1)`가 호출 됐어요. n은 1입니다.
   - n은 1입니다! `n == 1` 조건이 참이 됐어요.(base case 도달)
   - 함수는 더이상 실행되지 않고, 바로 1을 반환합니다.
4. `factorial(1)` 결과를 반환(역순 계산 시작)
    - `factorial(1)`이 1을 `factorial(2)`에게 반환
5. `factorial(2)`계산 및 반환
    - 2 * `factorial(1)`을 계산하고 있었음. `factorial(1)`의 결과인 1을 받았음 -> 2 * 1 = 2를 계산해서 반환
    - `factorial(2)`는 2 * 1 = 2 결과값을 `factorial(3)`에게 반환
6. `factorial(3)`계산 및 최종 반환
    - 3 * `factorial(2)` `factorial(2)`의 결과인 2를 받았으므로, 3 * 2 = 6을 계산
7. `factorial(3)` 6을 반환하고 끝난다.
