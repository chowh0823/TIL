```python
str_1 = "Python"
str_2 = "Python"

print(str_1 == str_2)  # True
print(str_1 is str_2)  # True -> 참조하고 있는 메모리 주소가 같다. 불변형을 참조

a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)  # True
print(a is b)  # False -> 참조하고 있는 메모리 주소가 다르다. 가변형을 참조

b = a  # 같은 주소를 참조

b[0] = 10
print(a[0])  # 같은 주소를 참조하기 때문에 b가 가리키고 있는 리스트 값이 바뀌면 a가 가리키고 있는 리스트 값이 바뀌는 것
```
